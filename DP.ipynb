**TRAIN MODEL**

```python
############## TRAIN MODEL ################
import os
import wandb
from dotenv import dotenv_values
from datetime import datetime
from mobilnet.experiment import DataModule
from mobilnet.model import EfficientUNet4Class
from mobilnet.train import Trainer


def run_experiment(cfg):
    dm = DataModule(
        train_dir=cfg["train_dir"],
        val_dir=cfg["val_dir"],
        test_dir=cfg["test_dir"],
        batch_size=cfg["batch_size"],
        stats_path=cfg["stats_json_path"],
    )

    class_weights = dm.get_class_weights()

    model = EfficientUNet4Class(
        in_channels=4,
        num_classes=4,
        use_domain_head=False,
        moddrop_nir_prob=cfg.get("moddrop_nir_prob", 0.5),
    )

    trainer = Trainer(
        model=model,
        datamodule=dm,
        cfg=cfg,
        class_weights=class_weights,
        log_dir=cfg["log_dir"],
    )

    trainer.fit(epochs=cfg["epochs"], early_stop_patience=6)
    trainer.test(num_plots=cfg.get("num_plots_test", 15))


if __name__ == "__main__":
    env_path = '/content/drive/MyDrive/Colab_Notebooks/config.env'
    config_values = dotenv_values(env_path)

    log_dir = os.path.join(
        '/content/drive/MyDrive/Colab_Notebooks/mobilnet_logs',
        datetime.now().strftime("%Y%m%d_%H%M%S"),
    )
    os.makedirs(log_dir, exist_ok=True)

    config = {
        "stats_json_path": "/content/drive/MyDrive/Colab_Notebooks/DP_code/data/kaggle/class_stats.json",
        "log_dir": log_dir,
        "train_dir": '/content/train',
        "val_dir": '/content/validation',
        "test_dir": '/content/test',
        "batch_size": 8,
        "epochs": 45,
        # --- HPO winners ---
        "lr": 0.0025,
        "weight_decay": 0.004,
        "focal_weight": 0.33,
        "label_smoothing": 0.05,
        "focal_gamma": 2.96,
        # "betas": [0.40, 0.75, 0.60, 0.80],
          "betas": [0.7280, 0.4572, 0.3002, 0.3221],
        "moddrop_nir_prob": 0.5,
        "num_plots_test": 15,
    }

    if config_values.get('WANDB_API_KEY'):
        os.environ['WANDB_API_KEY'] = config_values['WANDB_API_KEY']
        wandb.login()
        wandb.init(project="mobilnet-dp-fly", config=config)

    run_experiment(config)
```

```python
################## HOP #####################
import os
import random
import shutil
from mobilnet.hpo import run_hpo

def sample_hpo_subset(src_dir, dst_dir, fraction=0.10, seed=42):
    os.makedirs(dst_dir, exist_ok=True)
    all_files = sorted(os.listdir(src_dir))
    rgb_files = [f for f in all_files if "_RGB" in f.upper()]
    random.seed(seed)
    sampled = random.sample(rgb_files, int(len(rgb_files) * fraction))
    for f in sampled:
        shutil.copy(os.path.join(src_dir, f), dst_dir)
        base = f.upper().split("_RGB")[0]
        nir = next((n for n in all_files if n.upper().startswith(base + "_NIR")), None)
        if nir:
            shutil.copy(os.path.join(src_dir, nir), dst_dir)
    print(f"Copied {len(sampled)} pairs to {dst_dir}")

sample_hpo_subset("/content/train",      "/content/hpo_train", fraction=0.20)
sample_hpo_subset("/content/validation", "/content/hpo_val",   fraction=0.20)

best = run_hpo(
    train_dir="/content/hpo_train",
    val_dir="/content/hpo_val",
    test_dir="/content/test",
    stats_path="/content/class_stats.json",
    log_dir="/content/hpo_logs",
    n_trials_phase1=20,
    n_trials_phase2=15,
    epochs_per_trial=12,
    batch_size=8,
)

print("Done. Results saved to hpo_logs/hpo_results.json")
print(best["final_training_config"])
```
